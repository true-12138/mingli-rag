```python
df.describe()
```

```python
low_df_punc=df[df["punc_ratio"]<0.05]
print(low_df_punc[["filename","punc_ratio"]])
```

```python
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

plt.figure(figsize=(8,4))
plt.hist(df["punc_ratio"], bins=30)
plt.title("古籍标点占比分布")
plt.xlabel("标点占比")
plt.ylabel("文件数量")
plt.grid(alpha=0.3)
plt.show()
```
